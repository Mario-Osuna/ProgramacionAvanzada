```python
#Ejercicio 1

def is_integer(texto: str) -> bool:
    """Esta funcion recibe un texto el cual lo conviete a numero entero y retorna True si es posible la conversion"""
    try:
        int(texto)
        return True #Prueba a convertir el texto a un numero entero y retorna True
    except ValueError:
        return False #Si no es posible convertir el texto a un numero entero, retorna False

def seconds_to_dhms(total_seconds: int) -> tuple[int, int, int, int]:
    """Esta funcion recibe un numero entero el cual representa segundos y retorna una tupla con los dias, horas, minutos y segundos"""
    dias: int = (total_seconds // 86400)
    dias_rest: int = (total_seconds % 86400)
    horas: int = (dias_rest // 3600)
    horas_rest: int = (dias_rest % 3600)
    minutos: int = (horas_rest // 60)
    segundos: int = (horas_rest % 60)
    return dias, horas, minutos, segundos

def main() -> None:
    while True:
        raw_seg = input("Introduzca los segundos: ")    #Introduce los segundos y lo almacenamos en una variable
        if is_integer(raw_seg) and int(raw_seg) >=0:    #Comprobamos que los numeros introducidos son positivos y enteros
            total_seconds = int(raw_seg)                #Si cumple, convertimos los segundos introducidos por teclado en integer y los almacenamos
            break
        else:
            print("Número no válido, introduce un número entero y positivo") #Si no cumple, damos mensaje de error y volvemos al bucle

    dias, horas, minutos, segundos = seconds_to_dhms(total_seconds)
    print(f"{total_seconds} son {dias} d, {horas} h, {minutos} m y {segundos} s")

main()
```

```python
#Ejercicio 2

def is_leap_year(year: int) -> bool:
    """
    Devuelve True si la el año es bisiesto y devuelve un False en caso contrario
    Como es una expresion booliana, devuelve directamente un booliano
    """
    return (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)

def  leap_years_between(start: int, end: int) -> list[int]:
    bisiestos: list[int]=[] #Creamos una lista vacía para almacenar los numeros bisiestos
    for year in range (start, end + 1): #Recorremos toda la orquilla de numeros en busca de años bisiestos
        if is_leap_year(year): #Empleamos la funcion anterior para verificar si es bisiesto
            bisiestos.append[year] #Con .append lo agregamos a la lista
    return bisiestos

def explain_leap_year(year: int) -> str:
    if year % 400 == 0:
        print(f"{year} es bisiesto: es divisible entre 400")
    elif year % 100 == 0:
        print(f"{year} no es bisiesto: es divible entre 100 pero no entre 400")
    elif year % 4 == 0:
        print(f"{year} es bisiesto: es divisible entre 4 y no entre 100")
    else:
        print(f"{year} no es bisiesto: no es divisible entre 4")

def main() -> None:
    while True:
        year_str = input("Escribe un año: ")
        year = int(year_str)
        if year <= 1582:
            print("El calendario gregoriano aún no existia, debe de ser a partir de 1582")
            return False
        else:
            return True

    
```
